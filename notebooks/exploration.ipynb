## Load Data

```python
# Initialize loader
loader = CodeFrequencyLoader('../Code frequency.csv')
data = loader.load()

print(f"Loaded {len(data)} records")
data.head()
```

## Summary Statistics

```python
summary = loader.get_summary()

for key, value in summary.items():
    print(f"{key}: {value}")
```

## Analysis

```python
# Initialize analyzer
analyzer = CodeFrequencyAnalyzer(loader)

# Activity ratio
activity_ratio = analyzer.calculate_activity_ratio()
print(f"Activity Ratio: {activity_ratio:.2%}")
```

### Top Active Weeks

```python
top_weeks = analyzer.get_top_activity_weeks(10)
top_weeks
```

### Yearly Statistics

```python
yearly = analyzer.get_yearly_stats()
yearly
```

### Churn Statistics

```python
churn = analyzer.get_churn_stats()
churn
```

### Coding Sprints

```python
sprints = analyzer.detect_sprints()

print(f"Found {len(sprints)} coding sprints:\n")
for i, sprint in enumerate(sprints, 1):
    print(f"Sprint {i}:")
    for key, value in sprint.items():
        print(f"  {key}: {value}")
    print()
```
